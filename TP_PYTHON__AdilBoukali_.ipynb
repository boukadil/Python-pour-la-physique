```python
def position(t, x0=0.0, v0=0.0, a=9.81):
    """Position d'un mobile en MRUA (unités SI)."""
    return x0 + v0 * t + 0.5 * a * t**2

print(position(2.0))               # → 19.62  (chute libre)
print(position(2.0, v0=5.0))       # → 29.62
print(position(2.0, x0=10, a=0.0)) # → 10.0   (mouvement uniforme)
```

```python
def derivee(f, x, h=1e-6):
    """Dérivée numérique de f en x (différence centrée)."""
    return (f(x + h) - f(x - h)) / (2 * h)

def carre(x):
    return x**2

derivee(carre, 3.0)   # → ≈ 6.0
```

```python
def qsort(lst):
    """Tri rapide : renvoie une copie triée de la liste."""
    if len(lst) <= 1:                 # cas de base : déjà triée
        return lst
    pivot, reste = lst[0], lst[1:]    # 1er élément = pivot

    inferieurs = [x for x in reste if x < pivot]
    superieurs = [x for x in reste if x >= pivot]
    return qsort(inferieurs) + [pivot] + qsort(superieurs)
```

```python
print(qsort([5, 2, 9, 1, 7, 3]))
assert qsort([1, 4, 2, 5, 3]) == sorted([1, 4, 2, 5, 3])
print("Tests réussis")
```

```python
def vitesse(distance, duree):
    """Vitesse moyenne (m/s) pour une distance (m) et une durée (s)."""
    if duree <= 0:
        raise ValueError("La durée doit être strictement positive.")
    return distance / duree
```

```python
for d, t in [(100, 9.58), (100, 0)]:
    try:
        print(f"v = {vitesse(d, t):.2f} m/s")
    except ValueError as e:
        print("Erreur :", e)
```

# Carnet 4 — Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.linspace(-np.pi, np.pi, 256)
C = np.cos(X)
S = np.sin(X)

plt.figure(figsize=(8, 5))
plt.plot(X, C, color="blue", linewidth=2.5, linestyle="--", label="cosinus")
plt.plot(X, S, color="green", linewidth=2.5, label="sinus")
plt.xlabel("angle (rad)")
plt.ylabel("amplitude")
plt.title("Fonctions trigonométriques")
plt.legend(loc="upper left")
plt.grid(True)
plt.show()
```

```python
rng = np.random.default_rng(1)
t_mes = np.linspace(0, 5, 10)
x_mes = 2.0 * t_mes + 1.0 + rng.normal(0, 0.4, t_mes.size)
u_x = 0.4 * np.ones_like(t_mes)

plt.errorbar(t_mes, x_mes, yerr=u_x, fmt='o', capsize=3, label="mesures")
plt.plot(t_mes, 2.0 * t_mes + 1.0, '-', label="modèle")
plt.xlabel("t (s)")
plt.ylabel("x (m)")
plt.legend()
plt.show()
```

```python
mesures = rng.normal(loc=9.81, scale=0.15, size=500)
plt.hist(mesures, bins=20, edgecolor='white')
plt.axvline(mesures.mean(), color='r', label=f"moyenne = {mesures.mean():.2f}")
plt.xlabel("g (m/s²)")
plt.ylabel("effectif")
plt.legend()
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
ax1.plot(X, C)
ax1.set_title("cosinus")
ax2.hist(mesures, bins=20)
ax2.set_title("histogramme")
fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(mesures, bins=20, edgecolor='white')
ax.set_xlabel("g (m/s²)")
ax.set_ylabel("effectif")
fig.savefig("histogramme.png", dpi=300, bbox_inches='tight')
fig.savefig("histogramme.pdf", bbox_inches='tight')
plt.show()
```
